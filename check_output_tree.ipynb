```python
import uproot

# 替換為你的輸出文件路徑
file_path = "skimmed_nano_mc.root"

# 打開 ROOT 文件
with uproot.open(file_path) as file:
    # 列出文件中所有對象（例如樹、目錄）
    print("文件內容：", file.keys())
    
    # 假設樹名稱為 "Events"（NanoAOD 標準）
    tree = file["Events"]
    
    # 列出樹中的所有分支
    print("\n分支列表：")
    for branch in tree.keys():
        print(f" - {branch}: {tree[branch].interpretation}")
    
    # 可選：檢查前幾個事件的數據
    print("\n前 5 個事件的數據（示例分支）：")
    # 替換為你感興趣的分支名稱，例如 "Muon_pt"
    data = tree.arrays(["Muon_pt", "Jet_pt"], entry_stop=5)
    print(data)
```
